### Part 1.1 — A single neuron is just a line
A neuron computes `y_hat = w * x + b`. Fit it to the sine data with **manual gradient descent**.

```python
# TODO: Generate the toy data - DONE
#   x = np.linspace(-3, 3, 100); y = np.sin(x) + small Gaussian noise (std=0.1)
#   Plot it.

# Adding Gaussian noise + reshaping x to a column vector (useful for 1,2)
x = np.linspace(-3,3,100).reshape(-1, 1); y = np.sin(x) + np.random.normal(0, 0.1, size=x.shape) 
plt.plot(x,y)
plt.xlabel("x")
plt.ylabel("y")
plt.show()


# TODO: Initialise w and b to small random values. - DONE
w = np.random.randn() 
b = np.random.randn() 

# Creating a list to save the losses
losses = []

# TODO: Implement ONE gradient-descent step by hand, then loop it for ~200 steps: - DONE
#   y_hat = w * x + b
#   loss  = np.mean((y_hat - y) ** 2)               # MSE
#   dw    = np.mean(2 * (y_hat - y) * x)            # dLoss/dw  (derive this on paper!)
#   db    = np.mean(2 * (y_hat - y))                # dLoss/db
#   w    -= lr * dw
#   b    -= lr * db
#   Record the loss at every step.

# Initalise learning rate
lr = 0.01

# 200 step loop

for step in range(200):
    # Make predictions
    y_hat = w * x + b

    # Measure the current loss
    loss = np.mean((y_hat - y) ** 2)

    # Find the gradients
    dw = np.mean(2 * (y_hat - y) * x)
    db = np.mean(2 * (y_hat - y))
 
    # Updating parameters (weight & bias)
    w  = w - lr * dw
    b = b - lr * db

    # Store loss
    losses.append(loss)

# Ploting the loss curve
plt.plot(losses)
plt.xlabel("Iteration")
plt.ylabel("MSE Loss")
plt.title("Loss during gradient descent")
plt.show()


# Calculating the final fitted values
y_fit = w * x + b

# Plotting data and fitted line
plt.scatter(x, y, label="Actual data")
plt.plot(x, y_fit, label="Fitted line")
plt.xlabel("x")
plt.ylabel("y")
plt.title("Linear Regression Fit")
plt.legend()
plt.show()



# TODO: Plot (a) the loss curve over the 200 steps, and - DONE
#             (b) the data with the fitted line on top.
```

### Part 1.2 — Hidden layers and activations: why depth helps
Now build a tiny network **by hand**: 1 input -> a hidden layer of 8 neurons with `tanh` ->
1 output. Still NumPy, still manual gradients.

```python
# TODO: Initialise parameters - DONE
#   W1: shape (1, 8),  b1: shape (8,)
#   W2: shape (8, 1),  b2: shape (1,)
#   Small random values (e.g. np.random.randn(...) * 0.5)

W1 = np.random.randn(1, 8) * 0.1
b1 = np.zeros((8,))

W2 = np.random.randn(8, 1) * 0.1
b2 = np.zeros((1,))

# Losses list
losses = []

# TODO: Forward pass (keep the intermediate values — you need them for backprop): - DONE
#   z1 = x @ W1 + b1        # pre-activation, shape (100, 8)
#   h  = np.tanh(z1)        # activation
#   y_hat = h @ W2 + b2     # output, shape (100, 1)
#   loss  = np.mean((y_hat - y) ** 2)

# TODO: Backward pass (the chain rule, layer by layer): - DONE
#   d_yhat = 2 * (y_hat - y) / len(y)
#   dW2 = h.T @ d_yhat ;  db2 = d_yhat.sum(axis=0)
#   dh  = d_yhat @ W2.T
#   dz1 = dh * (1 - np.tanh(z1) ** 2)     # derivative of tanh
#   dW1 = x.T @ dz1 ;  db1 = dz1.sum(axis=0)

# TODO: Update all parameters with learning rate lr, loop for ~3000 steps, - DONE
#   record the loss.

# Training loop with tanh
for epoch in range(3000):
    # Forward pass
    z1 = x @ W1 + b1  # Pre-activation shape (100, 8)
    h = np.tanh(z1) # Activation 
    y_hat = h @ W2 + b2 # output, shape (100, 1)


    loss = np.mean((y_hat - y) ** 2) # loss (MSE)

    # Backward pass (Backpropagation)
    d_yhat = 2 * (y_hat - y) / len(y)
    dW2 = h.T @ d_yhat ; db2 = d_yhat.sum(axis=0)
    dh = d_yhat @ W2.T
    dz1 = dh * (1 - np.tanh(z1) ** 2) # derivative of tanh
    dW1 = x.T @ dz1; db1 = dz1.sum(axis=0)

    # Updating parameters
    W2 = W2 - lr * dW2
    b2 = b2 - lr * db2
    W1 = W1 - lr * dW1
    b1 = b1 - lr * db1

    # Recording losses
    losses.append(loss)

# Loss curve plot
plt.plot(losses)
plt.xlabel("Iteration")
plt.ylabel("MSE Loss")
plt.title("Loss with tanh")
plt.show()


# Plotting the final fit over the data (with tanh)

# Another forward pass (Final forward pass)
z1_fit = x @ W1 + b1
h_fit = np.tanh(z1_fit)  
y_fit = h_fit @ W2 + b2 


# Plotting (with ravel for flattening)
plt.scatter(x.ravel(), y.ravel(), label="Actual data")
plt.plot(x.ravel(), y_fit.ravel(), color="red", label="Neural network fit (tiny)")
plt.title("With tanh (activation)")
plt.legend()
plt.show()


# Holds losses without activation
losses_no_tanh = []


# Training loop without tanh
for epoch in range(3000):

    # Forward pass (without the activation (tanh))
    z1_no_h = x @ W1 + b1
    h = z1_no_h
    y_fit_no_h = h @ W2 + b2

    # Loss
    loss_no_tanh = np.mean((y_fit_no_h - y) ** 2)

    # Backward pass (Backpropagation - without tanh)
    d_yhat = 2 * (y_fit_no_h - y) / len(y)
    dW2 = h.T @ d_yhat ; db2 = d_yhat.sum(axis=0)
    dh = d_yhat @ W2.T
    dz1 = dh 
    dW1 = x.T @ dz1; db1 = dz1.sum(axis=0)

    # Updating parameters
    W2 = W2 - lr * dW2
    b2 = b2 - lr * db2
    W1 = W1 - lr * dW1
    b1 = b1 - lr * db1

    # Recording losses
    losses_no_tanh.append(loss_no_tanh)

# Loss curve plot (without tanh)
plt.plot(losses_no_tanh)
plt.xlabel("Iteration")
plt.ylabel("MSE Loss")
plt.title("Loss without tanh")
plt.show()

# Another forward pass before fit (Final without tanh)
z1_no_h = x @ W1 + b1
h = z1_no_h
y_fit_no_h = h @ W2 + b2

# Plotting (with ravel for flattening)
plt.scatter(x.ravel(), y.ravel(), label="Actual data")
plt.plot(x.ravel(), y_fit_no_h.ravel(), color="red", label="trained weights without activation (tanh)")
plt.title("Without tanh (activation)")
plt.legend()
plt.show()


# TODO: Plot the loss curve AND the final fit over the data. - DONE
#   Then re-run with the tanh REMOVED (h = z1). What happens to the fit?
```

### Part 1.3 — The learning rate: too cold, too hot, just right
The single most important hyperparameter. See its three regimes with your own eyes.

```python
# TODO: Wrap your Part 1.2 training loop in a function train_toy(lr, steps=3000) - DONE
#   that returns the list of losses.
def train_toy(lr: float, steps=3000) -> list[float]:

    np.random.seed(RANDOM_STATE)

    # Fresh parameters for every function call
    W1 = np.random.randn(1, 8) * 0.1
    b1 = np.zeros(8)

    W2 = np.random.randn(8, 1) * 0.1
    b2 = np.zeros(1)

    losses = [] # Losses list

    for epoch in range(steps):

        # Forward pass
        z1 = x @ W1 + b1  # Pre-activation shape (100, 8)
        h = np.tanh(z1) # Activation 
        y_hat = h @ W2 + b2 # output, shape (100, 1)


        loss = np.mean((y_hat - y) ** 2) # loss (MSE)

        # Backward pass (Backpropagation)
        d_yhat = 2 * (y_hat - y) / len(y)
        dW2 = h.T @ d_yhat ; db2 = d_yhat.sum(axis=0)
        dh = d_yhat @ W2.T
        dz1 = dh * (1 - np.tanh(z1) ** 2) # derivative of tanh
        dW1 = x.T @ dz1; db1 = dz1.sum(axis=0)

        # Updating parameters
        W2 = W2 - lr * dW2
        b2 = b2 - lr * db2
        W1 = W1 - lr * dW1
        b1 = b1 - lr * db1

        # Recording losses
        losses.append(loss)

    return losses

# TODO: Run it with THREE learning rates, e.g. lr = 0.001, 0.05, 1.0 - DONE
#   (keep the same initial weights for a fair comparison — reset the seed).
loss_0001 = train_toy(0.001)
loss_005 = train_toy(0.05)
loss_1 = train_toy(1.0)

# TODO: Plot the three loss curves on the same axes (log scale on y helps), with a legend. - DONE
plt.plot(loss_0001, label="lr = 0.001")
plt.plot(loss_005, label="lr = 0.05")
plt.plot(loss_1, label="lr = 1.0")
plt.xlabel("Iteration")
plt.ylabel("MSE Loss")
plt.title("Effect of the learning rate")
plt.yscale("log")
plt.legend()
plt.show()

```

```python
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler 


# TODO: Load the obesity dataset (same OBESITY_URL as Lab 2).
OBESITY_URL = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/GkDzb7bWrtvGXdPOfk6CIg/Obesity-level-prediction-dataset.csv"
obesity = pd.read_csv(OBESITY_URL)


# TODO: Encode categoricals exactly as you did in Lab 2
#   (binary yes/no -> 0/1; Gender/MTRANS/CAEC/CALC -> your justified choice).
#   Encode the target NObeyesdad into integers 0..6 (LabelEncoder) — CrossEntropyLoss
#   expects integer class indices.

# Making a copy
obesity_clean = obesity.copy()

# For binary columns
obesity_clean["family_history_with_overweight"] = obesity_clean["family_history_with_overweight"].map({
    "yes": 1,
    "no" : 0
})

obesity_clean["FAVC"] = obesity_clean["FAVC"].map({
    "yes":1,
    "no":0
})

obesity_clean["SMOKE"] = obesity_clean["SMOKE"].map({
    "yes":1,
    "no":0
})

obesity_clean["SCC"] = obesity_clean["SCC"].map({
    "yes":1,
    "no":0
})

#Ordinal-ish columns

# Making an ordinal map
ordinal_map = {
    "no":0,
    "Sometimes":1,
    "Frequently":2,
    "Always":3
}


# Mapping the columns in order (with the ordinal map)
obesity_clean["CAEC"] = obesity_clean["CAEC"].map(ordinal_map)
obesity_clean["CALC"] = obesity_clean["CALC"].map(ordinal_map)

# Nominal columns

obesity_clean["Gender"] = obesity_clean["Gender"].map({
    "Female" : 0,
    "Male" : 1
})


obesity_encoded = pd.get_dummies(
    obesity_clean,
    columns=["MTRANS"],
    drop_first=True
)

label_encoder = LabelEncoder() # Initalising the encoder

# Creating the domain feature (BMI)
obesity_encoded["BMI"] = obesity_encoded["Weight"] / (obesity_encoded["Height"] ** 2)


obesity_encoded["NObeyesdad_encoded"] = label_encoder.fit_transform(
    obesity_encoded["NObeyesdad"]
)


# TODO: Stratified train / validation / test split (60/20/20, stratify=y,
#   random_state=RANDOM_STATE).

# Separating X (features) and y (encoded NObeyesdad)
X = obesity_encoded.drop(columns=["NObeyesdad", "NObeyesdad_encoded"])
y = obesity_encoded["NObeyesdad_encoded"]


X_train_val, X_test, y_train_val, y_test = train_test_split(
    X,y, test_size=0.20, random_state=RANDOM_STATE, stratify=y
)

X_train, X_val, y_train, y_val = train_test_split(
    X_train_val, y_train_val, test_size=0.25, random_state=RANDOM_STATE, stratify=y_train_val
)


# TODO: Fit a StandardScaler on the TRAINING set only; transform all three splits.
#   (Neural networks are far more sensitive to feature scale than random forests — 
#    remember Section 1, where all gradients depended on x!)
numeric_columns = [
    "Age", "Height", "Weight", "FCVC", "NCP",
    "CH2O", "FAF", "TUE", "BMI"
]

scaler = StandardScaler()

X_train_scaled = X_train.copy()
X_val_scaled = X_val.copy()
X_test_scaled = X_test.copy()

X_train_scaled[numeric_columns] = scaler.fit_transform(X_train[numeric_columns])
X_val_scaled[numeric_columns] = scaler.transform(X_val[numeric_columns])
X_test_scaled[numeric_columns] = scaler.transform(X_test[numeric_columns])


# TODO: Convert to tensors and wrap in DataLoaders:
#   X -> torch.float32, y -> torch.long
#   train_loader (batch_size=32, shuffle=True), val_loader, test_loader

# Convert X and y to tensors and combine with TensorDataset in one go for train/validation/test
train_dataset = TensorDataset(torch.tensor(X_train.astype(float).values, dtype=torch.float32), torch.tensor(y_train.astype(int).values, dtype=torch.long))
val_dataset = TensorDataset(torch.tensor(X_val.astype(float).values, dtype=torch.float32), torch.tensor(y_val.astype(int).values, dtype=torch.long))
test_dataset = TensorDataset(torch.tensor(X_test.astype(float).values, dtype=torch.float32), torch.tensor(y_test.astype(int).values, dtype=torch.long))


# Creating the DataLoaders
train_loader = DataLoader(train_dataset, batch_size=32, shuffle=True)
val_loader = DataLoader(val_dataset, batch_size=32, shuffle=False)
test_loader = DataLoader(test_dataset, batch_size=32, shuffle=False)
```

### Part 2.2 — Design the network
An MLP: input -> hidden layers with ReLU -> 7 output logits.

```python
# TODO: Define a FeedForwardNet class (subclass nn.Module) with:
#   - constructor args: input_dim, hidden_sizes=(64, 32), n_classes=7
#   - at least TWO hidden layers with ReLU activations
#   - output layer producing 7 raw logits
#     (NO softmax — nn.CrossEntropyLoss applies log-softmax internally)

# class FeedForwardNet(nn.Module):
#     def __init__(self, ...):
#         ...
#     def forward(self, x):
#         ...

class FeedForwardNet(nn.Module):
    def __init__(self, input_dim, hidden_sizes=(64,32), n_classes=7):
        super(FeedForwardNet, self).__init__()

        # Layer 1
        self.fc1 = nn.Linear(input_dim, hidden_sizes[0])
        
        # Layer 2
        self.fc2 = nn.Linear(hidden_sizes[0], hidden_sizes[1])
        
        # Layer 3
        self.fc3 = nn.Linear(hidden_sizes[1], n_classes)

        # Activation function (ReLU)
        self.relu = nn.ReLU()

        def forward(self, x):
            out = self.fc1(x)
            out = self.relu(out)

            out = self.fc2(out)
            out = self.relu(out) # Activation is in between the hidden layers

            out = self.fc3(out)
            return out
        

# TODO: Instantiate, move to DEVICE, print the model.
model = FeedForwardNet(input_dim=X_train.shape[0]).to(DEVICE)


# TODO: Count trainable parameters:
#   sum(p.numel() for p in model.parameters() if p.requires_grad)
#   Verify the count for the FIRST layer by hand: input_dim * 64 + 64.

# Counting trainable parameters
trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
print(f"Total Trainable Parameters: {trainable_params}")

```
